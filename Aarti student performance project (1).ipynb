# 📊 Student Performance Analysis Report

## 🔹 1. Introduction

Student academic performance is influenced by multiple social, economic, and educational factors. 
Understanding these factors helps in improving learning outcomes and designing better educational policies.
This report analyzes a student performance dataset containing variables such as gender, parental education,
lunch type (economic condition), and test preparation course.
The objective of this study is to identify key factors affecting student scores in mathematics, reading, 
and writing, and to compare the findings with existing research studies.


    
## 🔹 2. Methodology

The analysis is based on a secondary dataset (commonly available on platforms like Kaggle).

## 🔹 3. Literature Review

Previous research highlights several factors influencing student performance.
Parental education plays a crucial role in academic success. William H. Jeynes (2002) found that
students with highly educated parents perform better due to academic support at home. This aligns
with the dataset findings.Socioeconomic status is another significant factor. Selcuk R. Sirin (2005) 
showed that students from higher economic backgrounds achieve better results. In this dataset, lunch 
type acts as an indicator of economic condition, where students with standard lunch perform better.
Test preparation has a strong positive effect on academic performance. Robert L. Bangert-Drowns et al.
(1983) demonstrated that coaching programs improve test scores. Similarly, students who completed the
preparation course scored higher.Gender differences are also observed. Janet Shibley Hyde (2005) found 
that females generally perform better in reading and writing, while males slightly outperform in 
mathematics. The dataset reflects this trend.Finally, reading and writing skills are closely related.
Timothy Shanahan (2006) identified a strong correlation between these skills, which is also evident in 
the dataset.

    
## 🔹 4. Key Findings

Students who completed test preparation scored significantly higher
Students with standard lunch performed better than those with free/reduced lunch
Higher parental education is associated with better student performance
Female students performed better in reading and writing
Male students performed slightly better in mathematics
Strong positive correlation exists between reading and writing scores



## 🔹 5. Conclusion

This study concludes that student performance is influenced by a combination of academic preparation, 
socioeconomic background, and parental education. The findings are consistent with existing research,
confirming that better resources and support systems lead to improved academic outcomes.
The dataset provides a realistic representation of real-world educational patterns. 
Future studies can apply machine learning models to predict student performance and identify
at-risk students for early intervention.

```python
import numpy as np

# Average score nikal rahe hain (real-life basis)
df['average'] = df[['math score', 'reading score', 'writing score']].mean(axis=1)

# Grade function
def get_grade(m):
    if m < 33:
        return 'F (Fail)'
    elif m < 50:
        return 'D'
    elif m < 65:
        return 'C'
    elif m < 80:
        return 'B'
    elif m < 90:
        return 'A'
    else:
        return 'A+'

# Apply grade on average
df['grade'] = df['average'].apply(get_grade)

df[['math score','reading score','writing score','average','grade']].head()
```
